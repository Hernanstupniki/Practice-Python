```python
# Stupniki Hernan
from pyspark.sql import SparkSession
from pyspark.sql.functions import col

spark = (
    SparkSession.builder
    .appName("ETL_act_01")
    .master("local[*]")
    .getOrCreate()
)
```

```python
# Primera actividad Construir un dataframe con los siguientes datos
df = spark.createDataFrame([
    {"Mes": "Enero", "Ventas ($)": 15000, "Temp(°C)": 18, "Publicidad ($)": 800, "Personal": 4, "Satisfacción": 4.2},
    {"Mes": "Feb", "Ventas ($)": 22000, "Temp(°C)": 25, "Publicidad ($)": 1200, "Personal": 5, "Satisfacción": 4.5},
    {"Mes": "Mar", "Ventas ($)": 18000, "Temp(°C)": 22, "Publicidad ($)": 900, "Personal": 4, "Satisfacción": 4.1},
    {"Mes": "Abr", "Ventas ($)": 28000, "Temp(°C)": 28, "Publicidad ($)": 1500, "Personal": 6, "Satisfacción": 4.8},
    {"Mes": "May", "Ventas ($)": 25000, "Temp(°C)": 30, "Publicidad ($)": 1300, "Personal": 5, "Satisfacción": 4.6},
])
df.show(truncate=False)
```

```python
# Calcular correlación entre temperatura y ventas
df.stat.corr("Temp(°C)", "Ventas ($)")
```

```python
# Identificar el mes con mejor retorno publicitario

df = df.withColumn(
    "Retorno_Publicidad",
    col("Ventas ($)") / col("Publicidad ($)")
)

df.orderBy(col("Retorno_Publicidad").desc()).show()
```

```python
# Analizar relación personal vs satisfacción cliente
df.stat.corr("Personal", "Satisfacción")
```

```python
print("Se propone mantener o replicar las estrategias publicitarias de los meses con mayor retorno, especialmente marzo, ya que presentó la mejor relación entre ventas e inversión. A su vez, sería conveniente reforzar la cantidad de personal en los períodos de mayor actividad, debido a que se observa una relación positiva entre el número de empleados y la satisfacción de los clientes. De esta manera, se buscaría mejorar tanto la rentabilidad de la publicidad como la calidad del servicio.")
```
